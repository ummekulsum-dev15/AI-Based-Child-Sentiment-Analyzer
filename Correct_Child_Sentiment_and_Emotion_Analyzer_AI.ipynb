This simple AI project uses TextBlob and Gradio to analyze the sentiment of a sentence.

```python
from textblob import TextBlob
import gradio as gr
```

```python
def analyze_sentiment(text):
    blob = TextBlob(text)
    polarity = blob.sentiment.polarity
    if polarity > 0:
        return "Sentiment: Positive"
    elif polarity < 0:
        return "Sentiment: Negative"
    else:
        return "Sentiment: Neutral"
```

```python
gr.Interface(fn=analyze_sentiment, inputs="text", outputs="text", title="Sentiment Analyzer").launch()
```
